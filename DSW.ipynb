## If we want to represent a group of values to a single entity where insertion order is preserved and duplicate values are allowed then we have to use list datatype

```python
#How to declare a list 
l = [] # empty list created 
print(type(l))
```

```python
l.append(10)
l.append(20)
l.append(30)
print(l)

while(True):
    num = int(input("Enter a number into list: "))
    l.append(num)

    dec = input("Do you want to continue: (Y/N): ")
    if dec == "Y":
        continue
    elif dec == "N":
        break
    else:
        print("Something else than Y or N")
        break
print(l)
```

```python
l.append('kris')
l
```

```python
l.append(None)
print(l)
```

```python
l[0] = 100
print(l)
```
