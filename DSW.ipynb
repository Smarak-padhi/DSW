```python
l.append(10)
l.append(20)
l.append(30)
print(l)

while(True):
    num = int(input("Enter a number into list: "))
    l.append(num)

    dec = input("Do you want to continue: (Y/N): ")
    if dec == "Y":
        continue
    elif dec == "N":
        break
    else:
        print("Something else than Y or N")
        break
print(l)
```

# SHALLOW VS DEEP COPY

Shallow copy and deep copy requires nested list

```python
import copy
x = [10,20,[30,40],50]
y = copy.copy(x) #copy.copy method creates two different object
x[0] = 888
print(x)
print(y)
```

```python
A shallow copy means creating new object (but not copying the inner nested object)
```
